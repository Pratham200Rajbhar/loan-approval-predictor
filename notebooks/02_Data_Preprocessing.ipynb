```python
import os
import pandas as pd
import numpy as np
import joblib
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder, StandardScaler
from sklearn.impute import SimpleImputer
import warnings

warnings.filterwarnings('ignore')
os.makedirs('data/processed', exist_ok=True)
os.makedirs('models', exist_ok=True)
```

```python
data = pd.read_csv('/home/pratham/Disk2/Projects/ml_projects/loan_prediction/data/raw/loan_data.csv')
```

```python
data
```

```python
data.drop('Loan_ID', axis=1, inplace=True)
```

```python
data
```

```python
data.info()
```

```python
cat_cols = ['Gender', 'Married', 'Dependents', 'Self_Employed']
num_cols = ['LoanAmount', 'Loan_Amount_Term', 'Credit_History']
```

```python
data.isnull().sum()
```

```python
# Impute Categorical Features
cat_imputer = SimpleImputer(strategy='most_frequent')
data[cat_cols] = cat_imputer.fit_transform(data[cat_cols])
```

```python
# Impute Numerical Features
num_imputer = SimpleImputer(strategy='median')
data[num_cols] = num_imputer.fit_transform(data[num_cols])
```

```python
data
```

```python
data.isnull().sum()
```

# Cell 5: Feature Engineering

```python
data['Dependents'] = data['Dependents'].replace('3+', 3).astype(int)

data['Total_Income'] = data['ApplicantIncome'] + data['CoapplicantIncome']

data['EMI'] = (data['LoanAmount'] * 1000) / data['Loan_Amount_Term']
```

```python
data[['ApplicantIncome', 'CoapplicantIncome', 'Total_Income', 'LoanAmount', 'Loan_Amount_Term', 'EMI']].head()
```

```python
data.info()
```

```python
encoders = ['Gender', 'Married', 'Education', 'Self_Employed', 'Property_Area', 'Loan_Status']
le_dict = {}
for col in encoders:
    le = LabelEncoder()
    data[col] = le.fit_transform(data[col])
    le_dict[col] = le
    
joblib.dump(le_dict, '/home/pratham/Disk2/Projects/ml_projects/loan_prediction/models/label_encoders.pkl')
```

```python
data.info()
```

```python
data
```

```python
x = data.drop('Loan_Status', axis=1)
y = data['Loan_Status']
```

```python
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42,stratify=y)
```

```python
x_train.shape, x_test.shape
```

```python
scaler = StandardScaler()

scale_cols = ['ApplicantIncome', 'CoapplicantIncome', 'LoanAmount', 'Loan_Amount_Term', 'Total_Income', 'EMI']

x_train[scale_cols] = scaler.fit_transform(x_train[scale_cols])
x_test[scale_cols] = scaler.transform(x_test[scale_cols])
```
